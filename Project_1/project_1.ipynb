Создадим функцию `create_dataset()` для открытия файлов с данными о мобильных приложениях ([AppleStore.csv](https://www.kaggle.com/datasets/ramamet4/app-store-apple-data-set-10k-apps) и [GooglePlayStore.csv](https://www.kaggle.com/datasets/lava18/google-play-store-apps)). Функция будет сохранять каждый файл данных в виде списка списков (list of lists):

```python
from csv import reader

def create_dataset(file, header=False):
    opened_file = open(file, encoding='utf8')
    read_file = reader(opened_file)
    data = list(read_file)
    if header:
        return data[1:], data[0]
        
    return data
```

Функция `create_dataset()` принимает два параметра:

- сам файл с данными
- указание, присутствует ли в файле заголовочная строка (header of columns names), содержащая в себе названия столбцов (по умолчанию `False`)

В случае, если header = `True`, т.е. заголовочная строка присутствует, функция возвращает:

- Список (list of lists) с данными по приложениям без заголовочной строки
- Отдельно заголовочную строку

Если же заголовочной строки нет, то функция просто возвращает данные в виде списка (list of lists)

```python
apps_dataset, apps_header = create_dataset('AppleStore.csv', header = True)
googleplay_dataset, googleplay_header = create_dataset('googleplaystore.csv', header = True)
```

На основе этих данных можно сформулировать критерий удаления дубликатов: для каждого приложения оставляем только запись с максимальным количеством отзывов, а остальные удаляем. Логика обоснована тем, что большее число отзывов обычно коррелирует с более актуальными данными.

Создадим функцию `max_reviews_dict()`, которая будет принимать набор данных и создавать словарь. В этом словаре ключами будут уникальные названия приложений, а значениями - максимальное количество отзывов для каждого приложения: 

```python
def max_reviews_dict(dataset, key_index, value_index):
    name_reviews_dict = {}
    for app in dataset:
        name = app[key_index]
        value = float(app[value_index])
        
        if name not in name_reviews_dict:
            name_reviews_dict[name] = value
        elif (name in name_reviews_dict) and (value > name_reviews_dict[name]):
            name_reviews_dict[name] = value

    return name_reviews_dict
```

Применим функцию `max_reviews_dict()` для определения максимального количества отзывов для каждого приложения в наборе данных Google Play:

```python
google_reviews_max = max_reviews_dict(googleplay_dataset, 0, 3) #Словарь с максимальным количеством отзывов по каждому приложению из Google Play
print(len(google_reviews_max))
```

Создадим функцию `eliminate_duplicate_lines()`, предназначенную для удаления повторяющихся строк из набора данных. Критерий удаления основан на максимальном количестве отзывов для каждого приложения - функция оставляет только те строки, где количество отзывов совпадает со значением из заранее созданного словаря. При этом гарантируется, что каждое приложение добавляется в новый очищенный набор данных только один раз:

```python
def eliminate_duplicate_lines(dataset, name_index, reviews_index, reviews_dict):
    clean_dataset = []
    already_added = []
    
    for app in dataset:
        name = app[name_index]
        reviews = float(app[reviews_index])
    
        if (reviews == reviews_dict[name]) and (name not in already_added):
            clean_dataset.append(app)
            already_added.append(name)
    
    return clean_dataset
```

## 2. ФИЛЬТРАЦИЯ ПРИЛОЖЕНИЙ
### (Отбор только бесплатных приложений с английскими названиями)

Создадим функцию `check_for_eng_characters()`, которая проверяет, содержит ли название приложения более трёх символов, не относящихся к английскому алфавиту (символы с кодом ASCII больше 127). Если таких символов больше трёх, функция возвращает `False`, в противном случае - `True`. Используется для определения, является ли название приложения в основном английским:

```python
def check_for_eng_characters(string):
    counter_non_eng_char = 0
    
    for character in string:
        if ord(character) > 127:
            counter_non_eng_char += 1
        if counter_non_eng_char > 3:
            return False
       
    return True
```

```python
print(check_for_eng_characters('Docs To Go™ Free Office Suite'))
print(check_for_eng_characters('Instachat 😜'))
print(check_for_eng_characters('爱奇艺PPS -《欢乐颂2》电视剧热播'))
```

Создадим функцию `eliminate_non_eng_apps()`, которая принимает датасет и индекс колонки с названием приложения. Внутри применяется созданная выше функция `check_for_eng_characters()` для фильтрации. В результате возвращается список приложений, содержащих только английские названия: 

```python
def eliminate_non_eng_apps(dataset, name_index):
    only_eng_apps = []
    
    for app in dataset:
        name = app[name_index]
        if check_for_eng_characters(name):
            only_eng_apps.append(app)
            
    return only_eng_apps
```

Для каждого датасета (Google Play и App Store) создадим отдельные списки, содержащие только те приложения, названия которых идентифицированы как преимущественно английские (по функции `check_for_eng_characters()`):

1) Для приложений из [Google Play](https://www.kaggle.com/datasets/lava18/google-play-store-apps):

```python
only_eng_googleplay_dataset = eliminate_non_eng_apps(googleplay_dataset, 0)
difference = len(googleplay_dataset) - len(only_eng_googleplay_dataset)
print('The number of Google Play apps with non-English names: ', difference)
```

Перед тем как выделить бесплатные приложения из обоих датасетов, нам необходимо изучить значения в столбцах, отвечающих за цену приложений. 

Для этого создадим функцию `freq_table()`, которая принимает датасет и индекс столбца, и возвращает словарь с уникальными значениями этого столбца и частотой их встречаемости в %. Это поможет нам понять, какие значения появляются в колонке с ценами, чтобы правильно задать условие для фильтрации:

```python
def freq_table(dataset, column_index):
    unique_values_dict = {}
    percentages_values_dict = {}
    
    for app in dataset:
        name = app[column_index]
        if name not in unique_values_dict:
            unique_values_dict[name] = 1
        else:
            unique_values_dict[name] += 1

    for value in unique_values_dict:
        percentages_values_dict[value] = round(unique_values_dict[value] / len(dataset) * 100, 2)
    
    return percentages_values_dict
```

```python
dict_googleplay = freq_table(only_eng_googleplay_dataset, 6)
print(dict_googleplay)
```

```python
dict_apps = freq_table(only_eng_apps_dataset, 4)
print(dict_apps)
```

Создадим функцию `eliminate_non_free_apps()`, которая принимает датасет, индекс колонки с ценой и значение, обозначающее бесплатное приложение (например, `Free` или `0.0`). Функция перебирает все приложения в датасете и добавляет в новый список только те, у которых цена соответствует заданному значению. В итоге функция возвращает список только с бесплатными приложениями:

```python
def eliminate_non_free_apps(dataset, price_index, free_price_value):
    only_free_apps = []
    
    for app in dataset:
        price = app[price_index] #Не переводим в тип float, т.к. по умолчанию все значения в списке являются строками
        if price == free_price_value:
            only_free_apps.append(app)
    
    return only_free_apps
```

Создадим функцию для сортировки `display_table()`, которая принимает на вход датасет и индекс столбца. Внутри она вызывает функцию `freq_table()`, которая возвращает словарь с процентным распределением значений заданного столбца (см. выше). Затем этот словарь преобразуется в список кортежей (list of tuples), где каждый кортеж содержит процент и соответствующее ему значение из столбца.

*Пример:* 

Было: {'Social Networking': 25}

Стало: [(25, 'Social Networking')\]

После этого список сортируется по процентам в порядке убывания. В конце функция построчно выводит отсортированные значения в выбранном столбце и их процентное соотношение.

```python
def display_table(dataset, column_index):
    percentages_dict = freq_table(dataset, column_index)
    percentages_list = []
    
    for value in percentages_dict: #преобразование словаря
        percentages_list.append((percentages_dict[value], value)) 
    
    sorted_percentages_list = sorted(percentages_list, reverse = True) 
    for entry in sorted_percentages_list:
        print(entry[1], ': ', entry[0])   
```

Далее рассмотрим среднее количество установок для каждого из жанров, чтобы понять какие из них наиболее популярны среди пользователей.

Для этой цели создадим две функции `correct_conversion_downloads()` и `avg_genre_downloads()`:

- `correct_conversion_downloads()` используется для проверки корректности формата строки, обозначающей количество скачиваний приложения. Так как мы будем преобразовывать строку в тип данных float, для дальнейшего вычисления среднего количества скачиваний, нам необходимо, чтобы строка корректно обрабатывалась. Исходя из анализа столбца `Installs` в наборе данных Google Play, данная строка задана в формате `100,000+`, что не сможет корректно перевестись в тип данных float.
  
- `avg_genre_downloads()` используется для вычисления среднего количества скачиваний приложения относительно каждого жанра. Внутри используется функция `correct_conversion_downloads()`, которая будет задействована для преобразования формата строки, если задан флаг `need_replacement` = `True`. В данном случае нам также необходимо передать список символов `replacements_list = [...]`, встечающихся в строке для их замены на '', то есть пустой символ. В результате из строки типа `100,000+` мы получим `100000`.

```python
def correct_conversion_downloads(string, list_of_replacements):
    correct_string = string
    
    for character in list_of_replacements:
        correct_string = correct_string.replace(character, '')
        
    return float(correct_string)    
```

```python
def avg_genre_downloads(dataset, genre_index, number_ratings_index, need_replacement = False, replacements_list = []):
    genres_dict = freq_table(dataset, genre_index)
    download_number_list = [] #список для дальнейшей сортировки
    
    for genre in genres_dict:
        total_ratings = 0
        len_genre = 0
        
        for app in dataset:
            genre_app = app[genre_index]
            
            if genre_app == genre:
                if need_replacement:
                    ratings_app = correct_conversion_downloads(app[number_ratings_index], replacements_list)
                else:
                    ratings_app = float(app[number_ratings_index])
                total_ratings += ratings_app
                len_genre += 1
                
        download_number_list.append((round(total_ratings / len_genre, 2), genre))
        
    sorted_download_number_list = sorted(download_number_list, reverse = True)  
    for entry in sorted_download_number_list:
         print(entry[1], ': ', entry[0])   
```
