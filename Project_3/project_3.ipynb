#### Рассмотрим распределение средних цен автомобилей в зависимости от их марки (`brand`):

Ограничимся топ-10 наиболее распространёнными брендами:

```python
brand_price_dict = {}

for a in autos['brand'].value_counts().index[:10]:
    selected_rows = autos.loc[autos['brand'] == a, 'price_usd']
    avg_price = selected_rows.mean()
    brand_price_dict[a] = avg_price

for key, value in brand_price_dict.items():
    print(f"{key}: ${round(value)}")
```

#### **Вывод**:

Среди 10 наиболее распространённых марок автомобилей наибольшая средняя цена наблюдается у премиальных брендов. Лидерами по стоимости являются `Audi ($9337)`, `Mercedes-Benz ($8628)` и `BMW ($8333)`. Это ожидаемый результат, поскольку данные марки относятся к премиальному сегменту рынка.

---

#### Для каждого из этих брендов вычислим среднее значение пробега (`odometer_km`):

```python
brand_mileage_dict = {}

for a in autos['brand'].value_counts().index[:10]:
    selected_rows = autos.loc[autos['brand'] == a, 'odometer_km']
    avg_mileage = selected_rows.mean()
    brand_mileage_dict[a] = avg_mileage

for key, value in brand_mileage_dict.items():
    print(f"{key}: {round(value)} mileage")
```

Для дальнейшего сравнения преобразуем словари в объекты pandas Series и объединим их в единый DataFrame для анализа средних значений пробега и цены по каждому бренду:

```python
brand_price_series = pd.Series(brand_price_dict)
brand_mileage_series = pd.Series(brand_mileage_dict)

df_brand_info = pd.DataFrame(brand_price_series, columns=['price_usd'])
df_brand_info['odometer_km'] = brand_mileage_series
```

```python
df_brand_info.sort_values('price_usd', ascending = False)
```

#### **Вывод**:

Несмотря на то, что автомобили марок `Audi ($9337)`, `Mercedes-Benz ($8628)` и `BMW ($8333)` имеют самые высокие средние цены, их средний пробег также находится среди наибольших в рассматриваемой группе и составляет около 129-133 тыс. км.

Это свидетельствует о том, что высокая стоимость автомобилей данных брендов не объясняется исключительно меньшим пробегом. Вероятно, на цену оказывают влияние и другие факторы, такие как класс автомобиля, технические характеристики, уровень оснащения, репутация бренда и восприятие его ценности на вторичном рынке.

Таким образом, в рассматриваемой выборке не наблюдается прямой зависимости между пробегом и стоимостью автомобиля: меньший пробег не обязательно соответствует более высокой цене.
